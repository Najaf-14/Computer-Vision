```python
import cv2
from google.colab.patches import cv2_imshow

path = "/content/video1.mp4"

cap = cv2.VideoCapture(path)

while cap.isOpened():
    ret, frame = cap.read()

    if not ret:
        break

    cv2_imshow(frame)

cap.release()
```

```python
import cv2
import os

image_folder = "/content/Images"
output_video = "/content/output.mp4"

# Get image files
images = [
    "image1.jpg",
    "image2.jpg",
    "image3.jpg",
    "image4.jpg"
]

# Read first image to get width and height
first_image = cv2.imread(os.path.join(image_folder, images[0]))

height, width, _ = first_image.shape

# Create video writer
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

video = cv2.VideoWriter(
    output_video,
    fourcc,
    1,              # FPS
    (width, height)
)

# Add images to video
for image_name in images:
    image_path = os.path.join(image_folder, image_name)

    frame = cv2.imread(image_path)

    video.write(frame)

# Release video
video.release()

print("Video created successfully:", output_video)
```

```python
import cv2
import os

image_folder = "/content/Images"
output_video = "/content/output.mp4"

# Get all image files
images = sorted([
    file for file in os.listdir(image_folder)
    if file.lower().endswith((".jpg", ".jpeg", ".png"))
])

print("Images found:", len(images))
print(images)

# Check if images exist
if len(images) == 0:
    print("No images found!")
else:

    # Read first image
    first_image = cv2.imread(
        os.path.join(image_folder, images[0])
    )

    height, width, _ = first_image.shape

    # Create video
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")

    video = cv2.VideoWriter(
        output_video,
        fourcc,
        1,                    # 1 image per second
        (width, height)
    )

    # Add all images
    for image_name in images:

        image_path = os.path.join(image_folder, image_name)

        frame = cv2.imread(image_path)

        # Resize if image has different dimensions
        frame = cv2.resize(frame, (width, height))

        video.write(frame)

    # Release video
    video.release()

    print("Video created successfully!")
    print("Saved at:", output_video)
```
