```python
import cv2
from google.colab.patches import cv2_imshow

image_path = '/content/Cartoon.jpg'
image = cv2.imread(image_path)

print("Orignal Image")
cv2_imshow(image)

center_coordinates = (100, 100)
radius = 30
color = (0,255,0)
thickness = 2

image_with_circle = cv2.circle(
    image,
    center_coordinates,
    radius,
    color,
    thickness
)

print("Circle")
cv2_imshow(image_with_circle)
```
