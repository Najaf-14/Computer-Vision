```python
import cv2
from google.colab.patches import cv2_imshow

image = '/content/Cartoon.jpg'
img = cv2.imread(image)

# Original ellipse
center = (200, 200)
axes = (100, 60)
angle = 0

cv2.ellipse(img, center, axes, angle, 0, 360, (0, 255, 0), 2)

# Convert ellipse to polygon
points = cv2.ellipse2Poly(
    center,
    axes,
    angle,
    0,
    360,
    100
)

# Draw polygon
cv2.polylines(img, [points], True, (255, 0, 0), 2)

print("Original Ellipse + Polygon")
cv2_imshow(img)
```

```python
import cv2
from google.colab.patches import cv2_imshow

image = '/content/Cartoon.jpg'
img = cv2.imread(image)

center = (200, 200)
axes = (100, 60)
angle = 0

# Convert ellipse to polygon
points = cv2.ellipse2Poly(
    center,
    axes,
    angle,
    0,
    360,
    60
)

# Fill polygon
cv2.fillConvexPoly(img, points, (0, 255, 255))

print("Filled Convex Polygon")
cv2_imshow(img)
```

```python
import cv2
from google.colab.patches import cv2_imshow

image = '/content/Cartoon.jpg'
img = cv2.imread(image)

text = "OpenCV Lab"
font = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 1
thickness = 2

# Get text size
(text_width, text_height), baseline = cv2.getTextSize(
    text,
    font,
    font_scale,
    thickness
)

print("Text Width:", text_width)
print("Text Height:", text_height)
print("Baseline:", baseline)

# Draw text
cv2.putText(
    img,
    text,
    (50, 100),
    font,
    font_scale,
    (0, 255, 0),
    thickness
)

# Draw rectangle around text
cv2.rectangle(
    img,
    (50, 100 - text_height),
    (50 + text_width, 100 + baseline),
    (255, 0, 0),
    2
)

print("getTextSize()")
cv2_imshow(img)
```
