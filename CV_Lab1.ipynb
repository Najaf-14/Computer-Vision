```python
import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/Images/Bird.jpg")

if image is None:
    print("Image could not be loaded. Check the file path.")
else:
    print("orignal Image:")
    cv2_imshow(image)

    height, width = image.shape[:2]
    # print(height)
    # print(width)

    resized = cv2.resize(
        image,
        (height * 2, width * 2)
    )

    print("Resized Image:")
    cv2_imshow(resized)
```

```python
import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/Images/Cars.jpg")

if image is None:
    print("Image could not be loaded. Check the file path.")
else:
    print("Original Image:")
    cv2_imshow(image)

    blurred = cv2.GaussianBlur(
        image,
        (3, 3), #should be odd numbers
        0
    )

    print("Gaussian Blurred Image:")
    cv2_imshow(blurred)
```

```python
import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/Images/Parrot.jpg")

if image is None:
    print("Image could not be loaded. Check the file path.")
else:
    print("Original Image:")
    cv2_imshow(image)

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    print("Grayscale Image:")
    cv2_imshow(gray)
```
